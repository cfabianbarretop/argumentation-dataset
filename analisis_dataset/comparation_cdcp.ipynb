## Comparação dos dataset CDCP

```python
import pandas as pd
from tabulate import tabulate
```

```python
result_path = "./results/cdcp/"
```

```python
df_original = pd.read_csv(result_path + "CDCP_Grafo.csv")
df_minimal = pd.read_csv(result_path + "CDCP_Minimal_Grafo.csv")
```

```python
df_original
```

```python
df_minimal
```

```python
df_original["id_compare"] = df_original["id"].astype(str)

df_minimal["id_compare"] = (
    df_minimal["text_id"]
    .astype(str)
    .str.replace("CDCP_", "", regex=False)
)
```

```python
list_file_comp = ["id_compare", "nós", "n_nodes", "n_edges", "n_comp"]
# list_file_comp = ["id_compare", "n_nodes", "n_edges", "n_comp"]
df_comparacion = df_original[list_file_comp].merge(
    df_minimal[list_file_comp],
    on="id_compare",
    suffixes=("_original", "_minimal")
)

# df_comparacion
# df_comparacion.describe()
print(tabulate(df_comparacion.describe().reset_index(), headers='keys', tablefmt='rounded_outline', showindex=False))
```

```python
df_diferencias = df_comparacion[
    (df_comparacion["n_comp_original"] == 1) #& # -> 328
    # (df_comparacion["n_nodes_original"] == df_comparacion["n_nodes_minimal"]) &
    # (df_comparacion["n_edges_original"] != df_comparacion["n_edges_minimal"]) 
    # (df_comparacion["n_comp_original"] == df_comparacion["n_comp_minimal"]) # -> Todos coiciden con el numero de componentes dentro del grafo
    # (df_comparacion["nós_original"] != df_comparacion["nós_minimal"])
]

print(df_diferencias.shape)
df_diferencias
```

```python
print(tabulate(df_diferencias.describe().reset_index(), headers='keys', tablefmt='rounded_outline', showindex=False))
```

```python
df_analizar = df_diferencias[df_diferencias["n_nodes_original"] > 10]
df_analizar = df_analizar[["id_compare", "nós_original", "n_nodes_original", "n_edges_original", "n_comp_original"]]
print(tabulate(df_analizar.reset_index(), headers='keys', tablefmt='rounded_outline', showindex=False))
```
